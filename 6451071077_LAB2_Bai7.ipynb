# Khai báo thư viện

```python
print("NguyenDucTinh - 6451071077")

%matplotlib inline
import matplotlib
import matplotlib.pyplot as plt
import numpy as np

from sklearn import datasets, tree, metrics
from sklearn.model_selection import train_test_split
```

# Tải tập dữ liệu chữ số viết tay

```python
print("NguyenDucTinh - 6451071077")

# Tải tập dữ liệu digits từ scikit-learn
digits = datasets.load_digits()
```

# Trực quan hóa một số mẫu dữ liệu huấn luyện

```python
print("NguyenDucTinh - 6451071077")

_, axes = plt.subplots(1, 4)
images_and_labels = list(zip(digits.images, digits.target))
for ax, (image, label) in zip(axes, images_and_labels[:4]):
    ax.set_axis_off()
    ax.imshow(image, cmap=plt.cm.gray_r, interpolation='nearest')
    ax.set_title('Training: %i' % label)
plt.show()
```

# Chuyển đổi dữ liệu ảnh (Flatten từ 2D sang 1D)

```python
print("NguyenDucTinh - 6451071077")

n_samples = len(digits.images)
data = digits.images.reshape((n_samples, -1))
```

# Sử dụng thuật toán ID3

```python
print("NguyenDucTinh - 6451071077")

# Khởi tạo bộ phân lớp Cây quyết định (ID3 với criterion='entropy')
classifier = tree.DecisionTreeClassifier(criterion="entropy", random_state=33)
```

# Chia dữ liệu thành tập huấn luyện và kiểm thử (80% - 20%)

```python
print("NguyenDucTinh - 6451071077")

X_train, X_test, y_train, y_test = train_test_split(
    data, digits.target, test_size=0.2, shuffle=False
)

print("Số lượng mẫu huấn luyện:", len(X_train))
print("Số lượng mẫu kiểm thử:", len(X_test))
```

# Tiến hành xây dựng cây ID3

```python
print("NguyenDucTinh - 6451071077")

# Huấn luyện mô hình trên tập huấn luyện
classifier.fit(X_train, y_train)
```

# Dự đoán trên tập kiểm thử

```python
print("NguyenDucTinh - 6451071077")

# Dự đoán nhãn cho tập kiểm thử
predicted = classifier.predict(X_test)
```

# Trực quan hóa kết quả dự đoán thực tế

```python
print("NguyenDucTinh - 6451071077")

_, axes = plt.subplots(1, 4)
images_and_predictions = list(zip(digits.images[n_samples // 2:], predicted))
for ax, (image, prediction) in zip(axes, images_and_predictions[:4]):
    ax.set_axis_off()
    ax.imshow(image, cmap=plt.cm.gray_r, interpolation='nearest')
    ax.set_title('Prediction: %i' % prediction)
plt.show()
```

# In báo cáo đánh giá chi tiết (Classification Report)

```python
print("NguyenDucTinh - 6451071077")

print("Classification report for classifier %s: \n%s\n"
      % (classifier, metrics.classification_report(y_test, predicted)))

disp = metrics.ConfusionMatrixDisplay.from_predictions(y_test, predicted)
disp.figure_.suptitle("Confusion Matrix")
print("Confusion matrix: \n%s" % disp.confusion_matrix)

plt.show()
```

# Biểu diễn Ma trận nhầm lẫn (Confusion Matrix)

```python
print("NguyenDucTinh - 6451071077")

# Vẽ ma trận nhầm lẫn trực quan
disp = metrics.ConfusionMatrixDisplay.from_predictions(y_test, predicted)
disp.figure_.suptitle("Confusion Matrix - Digits Classification")
plt.show()
```
